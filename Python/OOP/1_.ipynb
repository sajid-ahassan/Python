###  Object-Oriented Programming (OOP)
* **class** -> The keyword used to define a new blueprint/template for objects.
* **__self__** -> Represents the specific instance of the class; used to access variables belonging to the class.
* **__init__** -> The "constructor" method that runs automatically when a new object is created.
* **__str__** -> A special method that defines what is returned when you `print()` an object.
* **super()** -> Allows you to call methods from a parent (base) class.
* **inheritance** -> The process where one class takes on the attributes and methods of another.

```python
# The Bank Account: Create a BankAccount class with attributes for owner and balance. Add methods for deposit(amount) and withdraw(amount). Ensure the withdrawal method prevents the balance from going below zero.

class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def deposit(self,amount):
        self.balance = amount
    def withdraw(self,amount):
        if(amount <= self.balance):
            self.balance -= amount
        else:
            print("Insufficient balance")
        
o1 = BankAccount("Sajid",0)
o1.deposit(10000)
print(o1.balance)
o1.withdraw(5000)
print(o1.balance)
o1.withdraw(6000)
```

```python
# Inheritance (The Animal Kingdom): Create a base class Animal with a method speak(). Create two subclasses, Dog and Cat, that override the speak() method to print "Woof" and "Meow" respectively.

class Animal:
    def speak(self):
        print("What animal")
class Dog(Animal):
    def speak(self):
        print("Woof")
class Cat:
    def speak(self):
        print("Meow")

c1 = Cat()
c1.speak()

d1 = Dog()
d1.speak()
```

```python
# The Rectangle Class: Create a class Rectangle that takes length and width. Include methods to calculate area() and perimeter().

class Rectangle:
    def __init__(self,Length,width):
        self.w = width
        self.l = Length
        
        def area():
            return self.w*self.l
        self.area = area()
        
    
    
a = Rectangle(20,4)
a.area
```

```python
# Magic Methods (__str__): Modify your BankAccount or Rectangle class to include the __str__ magic method so that when you print() the object, it returns a human-readable string like "Rectangle of 10x20".

class Rectangle:
    def __init__(self,Length,width):
        self.w = width
        self.l = Length
        
        def area():
            return self.w*self.l
        self.area = area()
        
    def __str__(self):
        return f"{self.l}x{self.w}"
    
a = Rectangle(20,4)
print(a)

```

###  Lists (Ordered Sequences)
* **list()** -> Type constructor that creates a list or converts another iterable to a list.
* **.append()** -> Adds a single element to the very end of the list.
* **.extend()** -> Merges another list (or iterable) into the current list.
* **.insert()** -> Adds an element at a specific index position.
* **.pop()** -> Removes and returns the element at a specific index (defaults to the last).
* **.remove()** -> Searches for and deletes the first occurrence of a specific value.
* **.sort()** -> Reorders the list items in place (ascending by default).
* **.reverse()** -> Flips the order of the list items in place.
* **sorted(var,reverse = ,key)** -> Flips the order of the list items in place.

```python
# The Chunking Challenge: Write a function chunk_list(my_list, n) that takes a list and an integer n, and breaks the list into smaller lists of size n.Example: [1, 2, 3, 4, 5], 2 $\rightarrow$ [[1, 2], [3, 4], [5]].

lst = list(range(20))
n = int(input("Enter max number of element"))

fls =[]
tls = []

for i in lst:
    tls.append(i);
    if(len(tls)==n):
        fls.append(tls.copy())
        tls.clear()
if(len(tls)<n and len(tls)>0):
    fls.append(tls)
print(fls)
```

```python
# The "Second Best" Finder: Given a list of integers that might contain duplicates, write a script to find the second largest number without using the .sort() method.

lst = [2,4,62,10,98,13,53,23,53,10,62,73,98]

mx = max(lst)

dif = 10000
ans = 0
for i in lst:
    if((mx-i)<dif and i != mx):
        ans = i
        dif = mx-i
print(ans)

print(sorted(list(set(lst)))[-2])
```

```python
# Rotating the Carousel: Write a function that "rotates" a list to the right by k steps.Example: [1, 2, 3, 4, 5] rotated by 2 becomes [4, 5, 1, 2, 3].


lst = list(range(1,21))

n = int(input())

lst.extend(lst[:n])
del lst[:n]
print(lst)
```

```python
# The Zero Mover: Given a list of numbers, move all the zeros to the end of the list while maintaining the relative order of the non-zero elements. Do this "in-place" (modify the original list).

lst = [2,1,4,0,2,0,5,3,0,8]
print(id(lst))
# lst = [i for i in lst if(i!=0)]
for i in range(0,len(lst)):
    if lst[i] == 0:
        del lst[i]
        lst.append(0)
print(lst)
print(id(lst))
        
```

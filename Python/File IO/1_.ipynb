### 📂 File I/O Cheat Sheet

**Core Keywords**
* **open()** -> Opens a file and returns a file object.
* **with** -> Starts a context manager to ensure the file closes automatically.
* **as** -> Assigns the opened file object to a variable name.

---

**Access Modes (The 'How')**
* **'r'** -> Opens for reading only (the default).
* **'w'** -> Opens for writing; creates the file or overwrites it completely.
* **'rb'** -> used to read binary file(image,videoes etc).
* **'wb'** -> used to write binary file(image,videoes etc).
* **'a'** -> Opens for appending; adds new text to the end of the file.

---

**File Methods (The 'Action')**
* **.read()** -> Pulls the entire file content into one string.
* **.readline()** -> Pulls just the next single line of text.
* **.write()** -> Writes a specific string into the file.
* **.tell()** -> Shows the current position (byte) of the file cursor.
* **.seek()** -> Moves the file cursor to a specific byte position.
* **.close()** -> Manually shuts the file connection.

---

**JSON (JavaScript Object Notation)**
* *Use Case:* Sharing data between different languages, web APIs, and config files.
* **import json** -> Must import the built-in module first.
* **json.dump(obj, file)** -> Converts a Python object and saves it into a file.
* **json.load(file)** -> Reads a JSON file and converts it back into a Python object.
* **json.dumps(obj)** -> Converts an object to a JSON **string** (not a file).
* **json.loads(string)** -> Converts a JSON **string** back into a Python object.

---

**Pickle (Python-Specific Binary)**
* *Use Case:* Saving complex Python objects (like custom classes) that JSON can't handle.
* **import pickle** -> Must import the built-in module first.
* **pickle.dump(obj, file)** -> Serializes an object into a file (requires mode `'wb'`).
* **pickle.load(file)** -> Deserializes an object from a file (requires mode `'rb'`).
* **Warning:** Never "unpickle" data from an untrusted source, as it can execute malicious code.

 ##### serialization using json module

```python
import json
d = {
    "name" : "sajid",
    'age' : 22
}
# serialization using json module

with open("demo.json",'w') as f:
    json.dump(d,f,indent = 4)


# deserialization
with open("demo.json",'r') as f:
    d = json.load(f)
    print(d)
    print(type(d))
```

#### Pickling can be used to download ML model in binary fomate

```python
import pickle

class Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
    
    
p = Person("sajid",18)

with open("obj.pkl",'wb') as f:
    pickle.dump(p,f)
```

```python
with open("obj.pkl", 'rb') as f:
    a = pickle.load(f)
    print(a.name)
    print(type(a.age))
    a.display_info()
```

###  Dictionaries (Key-Value Pairs)
* **dict()** -> Type constructor used to create a dictionary.
* **.keys()** -> Returns a view object of all the keys in the dictionary.
* **.values()** -> Returns a view object of all the values in the dictionary.
* **.items()** -> Returns a view of all key-value pairs as tuples.
* **.get()** -> Safely retrieves a value; returns `None` (or a default) if the key doesn't exist.
* **.update()** -> Merges another dictionary or key-value pairs into the current one.
* **.popitem()** -> Removes and returns the last inserted key-value pair.

```python
# The Inventory Updater: You have a dictionary stock = {'apples': 5, 'oranges': 10}. Write a function add_stock(inventory, item, quantity) that increases the count if the item exists, or adds the new item if it doesn't.

stock = {
    'apples':5,
    'oranges':10,
    'grapes':1001
    
}
n = int(input('Number of item'))
def add_stock(key,item):
    if key in stock:
        stock[key]+=item
    else:
        stock[key] = item
        
while n:
    inp = input().split()
    k,v = inp[0],int(inp[1])
    add_stock(k,v)
    n-=1
print(stock)
```

```python
# Grade Book Aggregator: Given a dictionary where keys are student names and values are lists of test scores, create a new dictionary that maps each student's name to their highest score only.

stu = {
    'Alice':[99,89,45],
    'Bob':[78,56,98],
    'Rebeca':[98,87,99]
}

new_stu = {k:max(v) for (k,v) in stu.items()}


n_s = {}
for k,v in stu.items():
    n_s[k] = max(v)
    
n_s,new_stu
```

###  Sets (Unique Collections)
* **set()** -> Type constructor used to create a set or remove duplicates from a list.
* **.add()** -> Inserts a single new element into the set.
* **.remove()** -> Deletes a specific element; raises an error if the element is missing.
* **.discard()** -> Deletes a specific element but does NOT error if it is missing.
* **.union()** -> Returns a new set containing all elements from both sets.
* **.intersection()** -> Returns a new set containing only elements found in both sets.
* **.difference()** -> Returns elements that are in the first set but not the second.

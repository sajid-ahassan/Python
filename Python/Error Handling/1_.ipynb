###  Error Handling & Exceptions Reference

**Handling Keywords**
* **try** -> Wraps the block of code you want to test for errors.
* **except** -> Defines the code that runs if a specific error occurs.
* **else** -> Defines code to run only if the try block was successful (no errors).
* **finally** -> Defines code that runs no matter what happened, used for cleanup.
* **raise** -> Manually forces a specific error to occur.
* **assert** -> Checks if a condition is True, otherwise stops the program with an error.

**Standard Error Types (Exceptions)**
* **SyntaxError** -> Raised when the code rules are broken (missing colons, etc.).
* **IndentationError** -> Raised when spaces or tabs are not aligned correctly.
* **NameError** -> Raised when you try to use a variable that hasn't been defined.
* **TypeError** -> Raised when an operation is used on the wrong data type (e.g., adding a string to an int).
* **ValueError** -> Raised when a function gets the right type but an inappropriate value (e.g., `int("abc")`).
* **IndexError** -> Raised when you try to access a list index that is out of range.
* **KeyError** -> Raised when you try to access a dictionary key that does not exist.
* **FileNotFoundError** -> Raised when trying to open a file that cannot be found.
* **ZeroDivisionError** -> Raised when attempting to divide a number by zero.
* **AttributeError** -> Raised when an object does not have the method or property you are calling.
* **ImportError** -> Raised when a module or library cannot be found or loaded.

###  Best Practice Example
````` python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("That wasn't a valid number!")
except ZeroDivisionError:
    print("You can't divide by zero!")
else:
    print(f"Success! Result is {result}")
finally:
    print("Process complete.")

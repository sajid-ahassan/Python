###  Loops & Control Flow
* **if** -> Checks a condition; executes the block if the condition is `True`.
* **elif** -> Short for "else if"; checks another condition if the previous ones were `False`.
* **else** -> Executes a block of code if all preceding conditions were `False`.
* **for** -> Iterates over a sequence (list, tuple, string) or range.
* **while** -> Repeats a block of code as long as a condition remains `True`.
* **break** -> Immediately terminates the current loop.
* **continue** -> Skips the rest of the current loop iteration and moves to the next one.
* **pass** -> A null statement used as a placeholder (does nothing).
* **range()** -> Generates a sequence of numbers, commonly used in `for` loops.

```python
#  FizzBuzz (The Classic): Write a loop that prints numbers from 1 to 20. If a number is divisible by 3, print "Fizz"; if by 5, print "Buzz"; if by both, print "FizzBuzz".

for i in range(1,21):
    if((i%3 == 0) and (i%5 == 0)):
        print("FizzBizz")
    elif i%3 == 0:
        print("Fizz")
    elif i%5 == 0:
        print("Bizz")
        

    
```

```python
# Password Validator: Create a while loop that asks a user for a password. If the password is shorter than 8 characters, tell them it's too short and ask again. If it meets the length, print "Access Granted."

while True:
    pas = input("Enter password ")
    dig = False
    alpha = False
    for i in pas:
        if (i>='0' and i<='9'):
            dig = True
    for i in pas:
        if ((i>='a' and i<='z') or (i>='A' and i<='Z')):
            alpha = True
        
    if len(pas)<8:
        print("Length atleast 8")
    elif dig == False:
        print("Atleast 1 dig needed")
    elif alpha == False:
        print("Atleast 1 alpha needed")
    else:
        print("Success")
        break
        
    
```

```python
# The Factorial Finder: Use a for loop to calculate the factorial of a given number $n$ (where $n! = n \times (n-1) \times \dots \times 1$).

n = int(input("Enter a Number"))
x = 1
for i in range(1,n+1):
    x*=i
print(x)
```

```python
# Nested Patterns: Use nested loops to print a right-angled triangle of asterisks like this:
#n = 4
# 1
# 121
# 12321
# 1234321


n = int(input())

for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end = "")
    for j in range(i-1,0,-1):
        print(j,end='')
    print()
```
